```python
#This is problem no.1365--> How Many Numbers Are Smaller Then The Current Number:
class Solution:
    def smallerNumbersThanCurrent(self, nums: List[int]) -> List[int]:
        ans= []
        for i in nums:
            c= 0
            for j in nums:
                if j<i:
                    c+=1
            ans.append(c)
        return ans
```
