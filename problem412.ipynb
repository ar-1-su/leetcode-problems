**This is a FizzBuzz program. Here is a explanarion-->>**

1. The code initialise by creating an arraylid named ans ( ans = [] ) to store the result of fizzbuzz.

2. Then we will move ahead and create a for loop in range of 1 to n ( representing a number for which fizzbuzz needs compute )

3. If STDIP(Standard Input) is multiple of both 3(i % 3 ==0) and 5(i % 5 ==0) then FizzBuzz is added to result list.

4. If STDIP(Standard Input) is multiple of 3(i % 3 ==0) then Fizz is added to result list.

5.  If STDIP(Standard Input) is multiple of 5(i % 5 ==0) then Buzz is added to result list.

6. If STDIP(Standard Input) is not a multiple of both 3 and 5 then default integer is added to result list.

7. The final result, which is a list of strings representing the FizzBuzz output for numbers from 1 to N, is returned.

```python
from typing import List

n= int(input("enter a number: "))

def fizzBuzz(n: int) -> List[str]:
        ans = []

        for i in range(1, n + 1):
            if i%3==0 and i%5==0:
                ans.append("FizzBuzz")
            elif i%3==0:
                ans.append("Fizz")
            elif i%5==0:
                ans.append("Buzz")
            else:
                ans.append(str(i))

        return ans

print(fizzBuzz(n))


```
